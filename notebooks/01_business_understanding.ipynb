# Projet : Segmentation des clients – Online Retail II

## 1. Compréhension du besoin métier

### 1.1. Contexte du projet

Dans un environnement commercial de plus en plus concurrentiel,
la compréhension du comportement des clients représente un enjeu
important pour les entreprises de commerce électronique.

Les entreprises disposent d'un volume important de données
transactionnelles, mais ces données ne sont pas toujours exploitées
pour comprendre les différents profils de clients.

Ce projet s'appuie sur le jeu de données Online Retail II, qui
contient l'historique des transactions d'une entreprise britannique
de vente en ligne entre décembre 2009 et décembre 2011.

L'objectif est d'exploiter ces données afin d'identifier des groupes
de clients présentant des comportements d'achat similaires.

### 1.2. Besoin métier

L'entreprise souhaite mieux connaître ses clients afin de pouvoir
adapter ses stratégies commerciales et marketing à leurs comportements.

Cependant, elle ne dispose pas nécessairement de catégories
prédéfinies permettant de distinguer les différents profils de clients.

Le besoin consiste donc à identifier automatiquement des groupes
de clients ayant des caractéristiques d'achat similaires.

Cette segmentation pourrait notamment aider l'entreprise à :

- Identifier ses clients les plus importants.
- Repérer les clients occasionnels.
- Détecter les clients qui n'ont pas acheté depuis longtemps.
- Adapter les campagnes marketing selon les profils identifiés.
- Améliorer la fidélisation et la relation client.

### 1.3. Problématique

Comment les algorithmes de clustering K-Means et de classification
ascendante hiérarchique (CAH) peuvent-ils permettre d'identifier
des segments pertinents et stables de clients à partir de leurs
comportements d'achat, tout en tenant compte de la qualité des
groupes obtenus et de la complexité computationnelle ?

L'étude cherchera également à déterminer quelle méthode est la
plus adaptée aux contraintes de taille des données, de stabilité
des résultats et d'affectation de nouveaux clients.

### 1.4. Objectifs du projet

Les objectifs principaux sont :

1. Comprendre et analyser les données transactionnelles.
2. Nettoyer les données et traiter les anomalies.
3. Construire des indicateurs décrivant le comportement des clients.
4. Appliquer l'algorithme K-Means pour effectuer une segmentation.
5. Appliquer la classification ascendante hiérarchique (CAH).
6. Évaluer et comparer les résultats des deux méthodes.
7. Étudier leur stabilité et leur complexité computationnelle.
8. Interpréter les segments obtenus selon leur intérêt métier.

```python
### 1.5. Nature du problème de Machine Learning

Ce projet relève de l'apprentissage non supervisé (Unsupervised Learning).

En effet, le jeu de données ne contient pas de variable cible
indiquant à quel segment appartient chaque client.

Nous allons donc utiliser des algorithmes de clustering afin de
regrouper les clients en fonction de leurs similarités.

Les deux méthodes étudiées sont :

- K-Means : un algorithme de partitionnement qui cherche à former
  des groupes autour de centroïdes.
- CAH : une méthode hiérarchique qui construit progressivement
  des groupes à partir des observations les plus similaires.

L'objectif n'est pas de prédire une catégorie connue à l'avance,
mais de découvrir une structure pertinente dans les données.
```

```python
### 1.6. Contraintes opérationnelles

L'étude doit prendre en considération plusieurs contraintes :

**Taille de la population :**
Le jeu de données contient plus d'un million de transactions.
Cependant, notre population d'étude sera constituée des clients
identifiables grâce à leur identifiant CustomerID.

**Unité d'analyse :**
L'unité d'analyse sera le client et non la transaction.
Les transactions seront donc agrégées afin de construire un
profil comportemental pour chaque client.

**Affectation de nouveaux clients :**
K-Means permet d'affecter un nouveau client au centroïde le plus
proche sans réentraîner immédiatement le modèle.

La CAH ne possède pas naturellement le même mécanisme d'affectation.
Nous étudierons donc les possibilités d'extension de cette méthode.

**Fréquence de réentraînement :**
La fréquence de réentraînement dépendra de l'évolution des
comportements d'achat et de la stabilité des segments.
Elle pourra être discutée à partir des résultats expérimentaux.

**Complexité computationnelle :**
La taille du jeu de données devra être prise en compte,
notamment pour la CAH, dont le coût en mémoire peut devenir
important lorsque le nombre d'observations augmente.
```

```python
### 1.7. Variables envisagées pour la segmentation

Nous envisageons d'utiliser l'approche RFM pour représenter
le comportement d'achat des clients.

Cette approche repose sur trois indicateurs :

- Récence (Recency) : nombre de jours écoulés depuis le dernier achat.
- Fréquence (Frequency) : nombre de factures distinctes associées
  aux achats du client.
- Montant (Monetary) : montant total des achats du client.

Ces indicateurs permettront de caractériser les clients selon
la récence, la fréquence et la valeur de leurs achats.

Leur calcul sera réalisé lors de la phase de préparation
et de construction des variables.
```
